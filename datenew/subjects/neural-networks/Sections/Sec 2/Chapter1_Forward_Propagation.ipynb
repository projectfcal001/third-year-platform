## Coding the forward propagation algorithm

In this exercise, you'll write code to do forward propagation (prediction) for your first
neural network.

Each data point is a customer. The first input is how many accounts they have, and the
second input is how many children they have. The model will predict how many
transactions the user makes in the next year. You will use this data throughout the
first 2 chapters of this course.

The input data has been pre-loaded as `input_data`, and the weights are available in a
dictionary called `weights`. The array of weights for the first node in the hidden layer
are in `weights['node_0']`, and the array of weights for the second node in the hidden
layer are in `weights['node_1']`.

The weights feeding into the output node are available in `weights['output']`.

NumPy will be pre-imported for you as `np` in all exercises.


````python
import numpy as np

# --- Data pre-loaded for you in the real exercise; filled in here from the course
#     slides ("Forward propagation code") so this cell actually runs ---
input_data = np.array([2, 3])
weights = {
    'node_0': np.array([1, 1]),
    'node_1': np.array([-1, 1]),
    'output': np.array([2, -1]),
}

# Calculate node 0 value: node_0_value
node_0_value = (input_data * weights['node_0']).sum()

# Calculate node 1 value: node_1_value
node_1_value = (input_data * weights['node_1']).sum()

# Put node values into array: hidden_layer_outputs
hidden_layer_outputs = np.array([node_0_value, node_1_value])

# Calculate output: output
output = (hidden_layer_outputs * weights['output']).sum()

# Print output
print(output)
````

## The Rectified Linear Activation Function

An "activation function" is a function applied at each node. It converts the node's
input into some output.

The rectified linear activation function (called ReLU) has been shown to lead to very
high-performance networks. This function takes a single number as an input, returning 0
if the input is negative, and the input if the input is positive.

Here are some examples:
```
relu(3) = 3
relu(-3) = 0
```


````python
def relu(input):
    '''Define your relu activation function here'''
    # Calculate the value for the output of the relu function: output
    output = max(input, 0)

    # Return the value just calculated
    return(output)

# Calculate node 0 value: node_0_output
node_0_input = (input_data * weights['node_0']).sum()
node_0_output = relu(node_0_input)

# Calculate node 1 value: node_1_output
node_1_input = (input_data * weights['node_1']).sum()
node_1_output = relu(node_1_input)

# Put node values into array: hidden_layer_outputs
hidden_layer_outputs = np.array([node_0_output, node_1_output])

# Calculate model output (do not apply relu)
model_output = (hidden_layer_outputs * weights['output']).sum()
model_output=relu(model_output)
# Print model output
print(model_output)
````

Both node values were already positive (5 and 1), so ReLU leaves them unchanged and the
output matches the previous cell. ReLU only makes a difference when a node's input would
otherwise be negative — see the extra examples near the end of this notebook for a case
where that happens.


## Applying the network to many observations/rows of data

You'll now define a function called `predict_with_network()` which will generate
predictions for multiple data observations, which are pre-loaded as `input_data`. As
before, weights are also pre-loaded. In addition, the `relu()` function you defined in
the previous exercise has been pre-loaded.


````python
# --- Example data: several customer rows (accounts, children), each one a data
#     point to run through the network defined above ---
input_data = [
    np.array([3, 5]),
    np.array([1, -1]),
    np.array([0, 0]),
    np.array([8, 4]),
]

weights = {
    'node_0': np.array([2, 4]),
    'node_1': np.array([4, -5]),
    'output': np.array([2, 7]),
}

# Define predict_with_network()
def predict_with_network(input_data_row, weights):

    # Calculate node 0 value
    node_0_input = (input_data_row*weights['node_0']).sum()
    node_0_output = relu(node_0_input)

    # Calculate node 1 value
    node_1_input = (input_data_row*weights['node_1']).sum()
    node_1_output = relu(node_1_input)

    # Put node values into array: hidden_layer_outputs
    hidden_layer_outputs = np.array([node_0_output, node_1_output])

    # Calculate model output
    input_to_final_layer = (hidden_layer_outputs*weights['output']).sum()
    model_output = relu(input_to_final_layer)

    # Return model output
    return(model_output)

# Create empty list to store prediction results
results = []
for input_data_row in input_data:
    # Append prediction to results
    results.append(predict_with_network(input_data_row, weights))

# Print results
print(results)
````

## Multi-layer neural networks

In this exercise, you'll write code to do forward propagation for a neural network with
2 hidden layers. Each hidden layer has two nodes. The input data has been preloaded as
`input_data`. The nodes in the first hidden layer are called `node_0_0` and `node_0_1`.
Their weights are pre-loaded as `weights['node_0_0']` and `weights['node_0_1']`
respectively.

The nodes in the second hidden layer are called `node_1_0` and `node_1_1`. Their weights
are pre-loaded as `weights['node_1_0']` and `weights['node_1_1']` respectively.

We then create a model output from the hidden nodes using weights pre-loaded as
`weights['output']`.


````python
# --- Example data: a single customer row, and weights for a 2-hidden-layer network ---
input_data = np.array([3, 5])

weights = {
    'node_0_0': np.array([2, 4]),
    'node_0_1': np.array([4, -5]),
    'node_1_0': np.array([-1, 2]),
    'node_1_1': np.array([1, 2]),
    'output': np.array([2, 7]),
}

def predict_with_network(input_data):
    # Calculate node 0 in the first hidden layer
    node_0_0_input = (input_data * weights['node_0_0']).sum()
    node_0_0_output = relu(node_0_0_input)

    # Calculate node 1 in the first hidden layer
    node_0_1_input = (input_data * weights['node_0_1']).sum()
    node_0_1_output = relu(node_0_1_input)

    # Put node values into array: hidden_0_outputs
    hidden_0_outputs = np.array([node_0_0_output, node_0_1_output])

    # Calculate node 0 in the second hidden layer
    node_1_0_input = (hidden_0_outputs*weights['node_1_0']).sum()
    node_1_0_output = relu(node_1_0_input)

    # Calculate node 1 in the second hidden layer
    node_1_1_input = (hidden_0_outputs*weights['node_1_1']).sum()
    node_1_1_output = relu(node_1_1_input)

    # Put node values into array: hidden_1_outputs
    hidden_1_outputs = np.array([node_1_0_output, node_1_1_output])

    # Calculate model output: model_output
    model_output = (hidden_1_outputs*weights['output']).sum()

    # Return model_output
    return(model_output)

output = predict_with_network(input_data)
print(output)
````

## Extra example 1 — A network with only *one* hidden layer

This is a self-contained extra example (not from the original exercise) to make the
"one hidden layer" case fully explicit and easy to compare against the deeper networks
below.

Setup:
- **3 inputs** per customer: number of accounts, number of children, and years as a
  customer.
- **One hidden layer with 2 nodes** (`node_0`, `node_1`), ReLU activation.
- **One output node** producing the predicted number of transactions.

We also pick weights so that one of the hidden nodes receives a **negative** input, to
show ReLU actually zeroing something out (unlike the all-positive example above).


````python
# One hidden layer, 2 nodes, 3 inputs per customer: (accounts, children, years_as_customer)
input_data_1layer = np.array([2, 3, 1])

weights_1layer = {
    'node_0': np.array([1, 1, 1]),     # will get a positive input
    'node_1': np.array([-4, 1, -2]),   # will get a negative input -> ReLU clips it to 0
    'output': np.array([3, -1]),
}

def predict_one_hidden_layer(input_data_row, weights):
    # Hidden layer: 2 nodes
    node_0_input = (input_data_row * weights['node_0']).sum()
    node_0_output = relu(node_0_input)

    node_1_input = (input_data_row * weights['node_1']).sum()
    node_1_output = relu(node_1_input)

    hidden_layer_outputs = np.array([node_0_output, node_1_output])

    # Output layer
    output = (hidden_layer_outputs * weights['output']).sum()
    return output, node_0_input, node_1_input, hidden_layer_outputs

prediction, n0_in, n1_in, hidden_outputs = predict_one_hidden_layer(input_data_1layer, weights_1layer)

print(f"node_0 input: {n0_in}  -> ReLU output: {hidden_outputs[0]}")
print(f"node_1 input: {n1_in}  -> ReLU output: {hidden_outputs[1]}  (negative input, clipped by ReLU)")
print(f"hidden layer outputs: {hidden_outputs}")
print(f"final prediction: {prediction}")
````

Notice `node_1`'s input is negative (`-4*2 + 1*3 + -2*1 = -7`), so ReLU clips its output
to exactly 0 — it contributes nothing to the final prediction for this particular
customer. This is the behavior ReLU is named for, and it's easy to miss in an
all-positive example like the one at the top of this notebook.


## Extra example 2 — A network with *three* hidden layers

This extends the two-hidden-layer example above by one more layer, following exactly the
same pattern: each layer's outputs become the next layer's inputs. Each hidden layer has
2 nodes, ReLU activation, same 2-input customer data (accounts, children) as the
two-hidden-layer example.


````python
input_data_3layers = np.array([3, 5])

weights_3layers = {
    # Hidden layer 1
    'node_0_0': np.array([2, 4]),
    'node_0_1': np.array([4, -5]),
    # Hidden layer 2
    'node_1_0': np.array([-1, 2]),
    'node_1_1': np.array([1, 2]),
    # Hidden layer 3
    'node_2_0': np.array([3, -1]),
    'node_2_1': np.array([1, 0.5]),
    # Output layer
    'output': np.array([2, 7]),
}

def predict_with_three_hidden_layers(input_data, weights):
    # ---- Hidden layer 1 ----
    node_0_0_input = (input_data * weights['node_0_0']).sum()
    node_0_0_output = relu(node_0_0_input)

    node_0_1_input = (input_data * weights['node_0_1']).sum()
    node_0_1_output = relu(node_0_1_input)

    hidden_0_outputs = np.array([node_0_0_output, node_0_1_output])

    # ---- Hidden layer 2 ----
    node_1_0_input = (hidden_0_outputs * weights['node_1_0']).sum()
    node_1_0_output = relu(node_1_0_input)

    node_1_1_input = (hidden_0_outputs * weights['node_1_1']).sum()
    node_1_1_output = relu(node_1_1_input)

    hidden_1_outputs = np.array([node_1_0_output, node_1_1_output])

    # ---- Hidden layer 3 ----
    node_2_0_input = (hidden_1_outputs * weights['node_2_0']).sum()
    node_2_0_output = relu(node_2_0_input)

    node_2_1_input = (hidden_1_outputs * weights['node_2_1']).sum()
    node_2_1_output = relu(node_2_1_input)

    hidden_2_outputs = np.array([node_2_0_output, node_2_1_output])

    # ---- Output layer ----
    model_output = (hidden_2_outputs * weights['output']).sum()

    return model_output, hidden_0_outputs, hidden_1_outputs, hidden_2_outputs

output_3layers, h0, h1, h2 = predict_with_three_hidden_layers(input_data_3layers, weights_3layers)

print(f"Hidden layer 1 outputs: {h0}")
print(f"Hidden layer 2 outputs: {h1}")
print(f"Hidden layer 3 outputs: {h2}")
print(f"Final prediction: {output_3layers}")
````

### Comparing all three network depths on the same customer

Same customer (`[3, 5]`), same weights for the layers they share — only the number of
hidden layers changes. This shows how the exact same forward-propagation pattern
(multiply, add, activate) just repeats one more time per extra layer, and how the
prediction shifts as the network gets deeper.


````python
same_customer = np.array([3, 5])

# 1 hidden layer (reuse weights_1layer's pattern isn't apples-to-apples since inputs
# differ in size -- so for a fair comparison we reuse the *2-input* node_0/node_1
# weights already defined for the "many observations" example above)
one_layer_weights = {
    'node_0': weights_3layers['node_0_0'],
    'node_1': weights_3layers['node_0_1'],
    'output': weights_3layers['output'],
}

def predict_one_layer(input_data, weights):
    node_0_output = relu((input_data * weights['node_0']).sum())
    node_1_output = relu((input_data * weights['node_1']).sum())
    hidden_outputs = np.array([node_0_output, node_1_output])
    return (hidden_outputs * weights['output']).sum()

pred_1 = predict_one_layer(same_customer, one_layer_weights)
pred_2 = predict_with_network(same_customer)                                   # 2 hidden layers, from earlier
pred_3, *_ = predict_with_three_hidden_layers(same_customer, weights_3layers)  # 3 hidden layers

print(f"1 hidden layer  -> prediction: {pred_1}")
print(f"2 hidden layers -> prediction: {pred_2}")
print(f"3 hidden layers -> prediction: {pred_3}")
````
